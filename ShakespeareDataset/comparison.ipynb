# Bigram vs GPT

```python
import matplotlib.pyplot as plt

bigram = [(0, 4.7305, 4.7241), (300, 2.811, 2.8249), (600, 2.5434, 2.5682), (900, 2.4932, 2.5088), (1200, 2.4863, 2.5035), (1500, 2.4665, 2.4921), (1800, 2.4683, 2.4936), (2100, 2.4696, 2.4846), (2400, 2.4638, 2.4879), (2700, 2.4738, 2.4911)]
gpt = [(0, 4.2221, 4.2306), (500, 1.7596, 1.9151), (1000, 1.3939, 1.6054), (1500, 1.2648, 1.5238), (2000, 1.1856, 1.4977), (2500, 1.1225, 1.4939), (3000, 1.0695, 1.4866), (3500, 1.0174, 1.5107), (4000, 0.9603, 1.5122), (4500, 0.9086, 1.5366), (4999, 0.8536, 1.5596)]

colors = {"Bigram": "#2563eb", "GPT": "#d97706"}
```

## Training and validation loss: Bigram vs GPT


```python
fig, ax = plt.subplots(figsize=(10, 6))

for name, record in [("Bigram", bigram), ("GPT", gpt)]:
    
    step, Tloss, Vloss = zip(*record)
    ax.plot(step, Tloss, marker="o",label=f"{name} — training", color=colors[name])
    ax.plot(step, Vloss, marker="s", markerfacecolor="white", label=f"{name} — validation", color=colors[name])

ax.set(title="Training and validation loss: Bigram vs GPT",xlabel="Training step", ylabel="Cross-entropy loss (lower is better)")
ax.grid(alpha=0.2)
ax.legend()
plt.show()
```

## Interpretation

GPT loss is lower but has slightly overfitted.
